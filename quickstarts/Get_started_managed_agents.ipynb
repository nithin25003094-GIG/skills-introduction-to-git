```
import pandas as pd

# Creating the DataFrame as given in the assignment
data = {
    'Student': ['Asha', 'Rahul', 'Neha', 'Karan', 'Priya'],
    'Course': ['Finance', 'Marketing', 'Finance', 'HR', 'Finance'],
    'Marks': [78, 65, 88, 55, 72],
    'Attendance': [92, 85, 95, 76, 89]
}

df_students = pd.DataFrame(data)

# a) Display the first 3 rows of the DataFrame
print("--- a) First 3 rows ---")
print(df_students.head(3))

# b) Display only the Student and Marks columns
print("\n--- b) Student and Marks columns ---")
print(df_students[['Student', 'Marks']])

# c) Find students who scored more than 70 marks
print("\n--- c) Students with Marks > 70 ---")
students_above_70 = df_students[df_students['Marks'] > 70]
print(students_above_70)

# d) Find the average marks of all students
avg_marks = df_students['Marks'].mean()
print(f"\n--- d) Average Marks: {avg_marks:.2f} ---")

# e) Find the student with the highest marks
highest_mark_row = df_students.loc[df_students['Marks'].idxmax()]
print(f"\n--- e) Student with highest marks: {highest_mark_row['Student']} ({highest_mark_row['Marks']} Marks) ---")
```

```
import pandas as pd

# Given employee sales dataset
data = {
    'Employee': ['Anil', 'Bhavna', 'Chirag', 'Deepa', 'Esha', 'Farhan'],
    'Department': ['Finance', 'Marketing', 'Finance', 'HR', 'Marketing', 'Finance'],
    'Sales': [45000, 38000, 52000, 30000, 42000, 48000]
}

sales_df = pd.DataFrame(data)

# a) Display the complete DataFrame
print("--- a) Complete DataFrame ---")
print(sales_df)

# b) Find the total sales generated by all employees
total_sales = sales_df['Sales'].sum()
print(f"\n--- b) Total Sales Generated: ₹{total_sales:,} ---")

# c) Find the average sales
average_sales = sales_df['Sales'].mean()
print(f"\n--- c) Average Sales: ₹{average_sales:,.2f} ---")

# d) Find the total sales department-wise using groupby()
dept_sales = sales_df.groupby('Department')['Sales'].sum().reset_index()
print("\n--- d) Department-wise Total Sales ---")
print(dept_sales)

# e) Identify the employee with the highest sales
top_employee = sales_df.loc[sales_df['Sales'].idxmax()]
print(f"\n--- e) Top Employee: {top_employee['Employee']} with Sales of ₹{top_employee['Sales']:,} (Dept: {top_employee['Department']}) ---")
```
